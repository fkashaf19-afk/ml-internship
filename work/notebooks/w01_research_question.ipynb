## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

**Decision:** Given a limited review budget (a content team can realistically review
~50 pages a week), which pages should be reviewed first — for refresh, expansion, or
pruning?

**Unit of analysis:** one content page (a `content_id`), scored and ranked
individually — not a client or a time-period aggregate.

**Who acts, and how:** A content editor opens the ranked queue and works top-down,
applying the suggested action tied to each page's reason code (e.g. rewrite
title/meta, refresh copy, monitor).

**Cost of a wrong call:**
- False positive (reviewed a page that wasn't a real problem): low cost — wasted
  editor time, deprioritized effort.
- False negative (missed a page that was genuinely declining with real demand):
  higher cost — lost traffic that compounds the longer it's ignored.

This asymmetry is why precision@K (not plain accuracy) is the right metric — it
matches how a reviewer actually works down a ranked list.
